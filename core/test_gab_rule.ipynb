#### Create fake images (box boundries) to test the gap_detection rule

```python
bounding_box = [
 [377.0, 164.4, 458.8, 237.1],  # 1
 [30.2,  52.4,  111.0, 124.1],  # 2
 [120.4, 278.9, 201.0, 353.2],  # 3
 [28.4,  163.2, 109.5, 236.7],  # 4
 [450.3, 51.7,  532.6, 122.8],  # 5
 [330.7, 277.3, 412.5, 350.4],  # 6
 [118.5, 50.9,  199.3, 121.6],  # 7
 [201.8, 165.5, 283.0, 238.0],  # 8
 [33.0,  276.5, 114.2, 351.0],  # 9
 [205.7, 54.1,  287.9, 126.0],  # 10
 [289.1, 162.0, 371.4, 235.2],  # 11
 [114.0, 160.8, 196.2, 234.1],  # 12
]
```

```python
bounding_box_sorted = sorted(
    bounding_box,
    key=lambda box: (box[1] + box[3]) / 2
)

```

```python
centre_y = [(box[1] + box[3]) / 2 for box in bounding_box_sorted]

print(centre_y)
```

```python

groups = [[]]

for j in range(len(centre_y)):

    if j > 0 and centre_y[j] - centre_y[j-1] > 100:
        groups.append([])

    groups[-1].append(bounding_box_sorted[j])

groups
```

```python
SEUIL = 100   # pixels

for numero, etagere in enumerate(groups, start=1):
    etagere_triee = sorted(etagere, key=lambda box: box[0])   # tri par x1
    for k in range(1, len(etagere_triee)):
        gauche = etagere_triee[k-1]
        droite = etagere_triee[k]
        ecart = droite[0] - gauche[2]                         # x1(droite) - x2(gauche)
        if ecart > SEUIL:
            print(f"Produit absent dans l'étagère {numero} ")
```

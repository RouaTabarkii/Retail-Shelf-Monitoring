```python
from pathlib import Path
```

```python
data = Path("/kaggle/input/datasets/thedatasith/sku110k-annotations")

total_size = sum(
    f.stat().st_size
    for f in data.rglob("*")
)

print(f"Dataset size: {total_size / (1024**3):.2f} GB")
```

```python
train_data = Path("/kaggle/input/datasets/thedatasith/sku110k-annotations/SKU110K_fixed/images/train")

total_size = sum(
    f.stat().st_size
    for f in train_data.rglob("*")
)

print(f"Training dataset size: {total_size / (1024**3):.2f} GB")
```

```python
test_data = Path("/kaggle/input/datasets/thedatasith/sku110k-annotations/SKU110K_fixed/images/test")

total_size = sum(
    f.stat().st_size
    for f in test_data.rglob("*")
)

print(f"Testing dataset size: {total_size / (1024**3):.2f} GB")
```

```python
val_data = Path("/kaggle/input/datasets/thedatasith/sku110k-annotations/SKU110K_fixed/images/val")

total_size = sum(
    f.stat().st_size
    for f in val_data.rglob("*")
)

print(f"Validation dataset size: {total_size / (1024**3):.2f} GB")
```

```python
test_label = Path("/kaggle/input/datasets/thedatasith/sku110k-annotations/SKU110K_fixed/labels/test")
train_label = Path("/kaggle/input/datasets/thedatasith/sku110k-annotations/SKU110K_fixed/labels/train")
val_label = Path("/kaggle/input/datasets/thedatasith/sku110k-annotations/SKU110K_fixed/labels/val")
```

### Validate data test : 

```python
list_image = []
for f in test_data.rglob("*.jpg") :
    list_image.append(f.stem)

```

```python
list_label = []

for f in test_label.rglob("*.txt") :
    list_label.append(f.stem)
```

```python
if sorted(list_label) == sorted(list_image) :
    print("they contain the same data")
else :
    print("different content ")
```
